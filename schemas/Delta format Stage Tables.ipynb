# Tables Schnema


###Creating Delta format Tables with SQL API


```python
spark.sql("""
CREATE TABLE IF NOT EXISTS stg_subscribers (
  txn_id STRING NOT NULL,
  member_id STRING NOT NULL,
  member_id_qual STRING DEFAULT '0F',
  ssn STRING,
  ssn_qual STRING DEFAULT '34',
  first_name STRING,
  last_name STRING,
  middle_name STRING,
  prefix STRING,
  suffix STRING,
  birth_date DATE,
  gender STRING,
  phone STRING,
  phone_qual STRING DEFAULT 'TE',
  address_line1 STRING,
  city STRING,
  state STRING,
  zip STRING,
  country STRING DEFAULT 'US',
  employment_status STRING,
  maint_type STRING,
  maint_date STRING,
  ingested_at TIMESTAMP,
  CONSTRAINT stg_subscribers_pk PRIMARY KEY (txn_id, member_id)
)
USING DELTA
TBLPROPERTIES('delta.feature.allowColumnDefaults' = 'supported')
""")
```

```python
spark.sql('''
          CREATE TABLE IF NOT EXISTS stg_dependents(
              txn_id STRING NOT NULL,
              subscriber_id STRING NOT NULL,
              member_id STRING NOT NULL,
              member_id_qual STRING DEFAULT '0F',
              relationship_code STRING,
              first_name STRING,
              last_name STRING,
              birth_date DATE,
              gender STRING,
              ssn STRING,
              ssn_qual STRING,
              main_type STRING,
              maint_date STRING,
              ingested_at TIMESTAMP,
              CONSTRAINT stg_depedents_pk PRIMARY KEY (txn_id)
          )
          USING DELTA
          TBLPROPERTIES('delta.feature.allowColumnDefaults' = 'supported')
          ''')
```
